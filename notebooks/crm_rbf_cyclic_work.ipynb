```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Input CSV
CSV_PATH = Path("/home/dysco/FLOWUnsteady/VarFLEXI-rVPM-Fenics/results/rbf_cyclic_error/coupling/coupling_history.csv")

# Directory in which the four PDF figures will be saved.
OUTPUT_DIR = CSV_PATH.parent
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Required columns for the four cyclic work round-trip diagnostics.
REQUIRED_COLUMNS = [
    "time",
    "fluid_force_round_trip_residual",
    "fluid_force_round_trip_relative_error",
    "solid_displacement_round_trip_residual",
    "solid_displacement_round_trip_relative_error",
]

if not CSV_PATH.exists():
    raise FileNotFoundError(f"CSV file not found: {CSV_PATH}")

coupling = pd.read_csv(CSV_PATH)
missing = [c for c in REQUIRED_COLUMNS if c not in coupling.columns]
if missing:
    raise KeyError(f"Missing required CSV columns: {missing}")

time = coupling["time"].to_numpy(dtype=float)

plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Helvetica", "Arial", "Liberation Sans"],
    "font.size": 7,
    "axes.labelsize": 8,
    "axes.titlesize": 8,
    "xtick.labelsize": 7,
    "ytick.labelsize": 7,
    "axes.linewidth": 0.5,
    "xtick.major.width": 0.5,
    "ytick.major.width": 0.5,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
})

print(f"Loaded {len(coupling)} coupling steps.")
print(f"Time range: {time.min():.6g} to {time.max():.6g} s")
```

```python
# 1. Fluid-force round-trip residual

y = coupling["fluid_force_round_trip_residual"].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(4.0, 2.4), dpi=300)
ax.plot(time, y, linewidth=1.2)

ax.set_xlabel(r"Time $t$ (s)", labelpad=4)
ax.set_ylabel("Fluid-force round-trip residual", labelpad=4)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(False)

fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fluid_force_round_trip_residual.pdf", bbox_inches="tight", transparent=True)
plt.show()
```

```python
# 2. Fluid-force round-trip relative error

y = coupling["fluid_force_round_trip_relative_error"].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(4.0, 2.4), dpi=300)
ax.plot(time, y, linewidth=1.2)

ax.set_xlabel(r"Time $t$ (s)", labelpad=4)
ax.set_ylabel("Fluid-force round-trip relative error", labelpad=4)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(False)

fig.tight_layout()
fig.savefig(OUTPUT_DIR / "fluid_force_round_trip_relative_error.pdf", bbox_inches="tight", transparent=True)
plt.show()
```

```python
# 3. Solid-displacement round-trip residual

y = coupling["solid_displacement_round_trip_residual"].to_numpy(dtype=float)

fig, ax = plt.subplots(figsize=(4.0, 2.4), dpi=300)
ax.plot(time, y, linewidth=1.2)

ax.set_xlabel(r"Time $t$ (s)", labelpad=4)
ax.set_ylabel("Solid-displacement round-trip residual", labelpad=4)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(False)

fig.tight_layout()
fig.savefig(OUTPUT_DIR / "solid_displacement_round_trip_residual.pdf", bbox_inches="tight", transparent=True)
plt.show()
```
